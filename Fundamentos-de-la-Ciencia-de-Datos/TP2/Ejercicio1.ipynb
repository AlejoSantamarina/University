```python
#Descarga de archivos de internet en mi PC.
import urllib.request
import os

url1 = "https://ignaciorlando.github.io/datasets/data-science/ObesityDataSet_raw_and_data_sinthetic.txt"
url2 = "https://ignaciorlando.github.io/datasets/data-science/ObesityDataSet_raw_and_data_sinthetic.csv"
carpeta_destino = "/home/alejo/Projects/University/Fundamentos-de-la-Ciencia-de-Datos/Archivos"

os.makedirs(carpeta_destino, exist_ok=True)

urllib.request.urlretrieve(url1, os.path.join(carpeta_destino, "descripcion.txt"))
urllib.request.urlretrieve(url2, os.path.join(carpeta_destino, "obesity.csv"))
```
