```python
#Descargar automaticamente ambos archivos
import urllib.request
import os

url1 = "https://ignaciorlando.github.io/datasets/data-science/netflix_titles.txt"
url2 = "https://ignaciorlando.github.io/datasets/data-science/netflix_titles.csv"
carpeta_destino = "/home/alejo/Projects/University/Fundamentos-de-la-Ciencia-de-Datos/Archivos"

os.makedirs(carpeta_destino, exist_ok = True)

urllib.request.urlretrieve(url1, os.path.join(carpeta_destino, "descripcion_2.txt"))
urllib.request.urlretrieve(url2, os.path.join(carpeta_destino, "archivo_2.csv"))
```

```python
#Lectura archivo .txt
ruta_txt = os.path.join(carpeta_destino, "descripcion_2.txt")

with open(ruta_txt, "r", encoding = "utf-8") as archivo:
    contenido = archivo.read()
    print(contenido)
```

```python
#Apertura de archivo .csv
ruta_csv = os.path.join(carpeta_destino, "archivo_2.csv")
with open(ruta_txt, "r", encoding = "utf-8") as archivo:
    print(f"Archivo {ruta_txt} abierto")
```

```python
#Impresión cabecera de archivo .csv
import pandas as pd
df = pd.read_csv(ruta_csv, encoding = "latin-1")
print(df.head)
```

```python
#Ver la cantidad de directores que hay.
total_directores = df["director"].nunique()
print(f"La cantidad de directores es de: {total_directores}")
```

```python
#Ver el director con mas peliculas
director_con_mas_peliculas = df["director"].value_counts().idxmax()
print(f"El director con mas peliculas es: {director_con_mas_peliculas}")
```

```python
#Contador de series y programas de TV
#No encuentro que hayan series, solo programas de TV y peliculas
cant_programaTV = (df["type"] == "TV Show").sum()
print(f"Hay un total de {cant_programaTV} programas de TV")
```

```python
#Impresión de los primeros 5 registros del archivo
import pandas as pd
df = pd.read_csv(ruta_csv, encoding = "latin-1")
print(df.head(5))
```

```python
#Dado un pais ver la cantidad de peliculas que tiene ese pais
nombre_pais = input("Escribe el nombre del pais")
cant_peliculas = (df["country"] == nombre_pais).sum()
print(f"El pais {nombre_pais} tiene {cant_peliculas} peliculas")
```

```python
#Buscar pelicula mas vieja

#Filtrar solo las peliculas
peliculas = df[df["type"] == "Movie"]

# Buscar el año mínimo dentro de esas películas
año_pelicula = peliculas["release_year"].min()

# Obtener el nombre de la película más vieja de ese año
nombre_pelicula = peliculas[peliculas["release_year"] == año_pelicula]["title"].iloc[0]

print(f"La película más vieja es {nombre_pelicula} del año {año_pelicula}")
```
