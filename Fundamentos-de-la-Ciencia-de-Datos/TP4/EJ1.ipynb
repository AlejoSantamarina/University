```python
#Descarga de datos.
import pandas as pd
ruta = "/home/alejo/Projects/University/Fundamentos-de-la-Ciencia-de-Datos/TP4/movies.csv"
df = pd.read_csv(ruta)
df.head()
#Solución para eliminar los valores NaN
df_limpio = df.dropna()
df_limpio.head()
```

```python
#Correlación entre recaudación(gross) y rating
correlación = df_limpio['RATING'].corr(df_limpio['Gross_Float'])
print(correlación)
```

```python
#Correlación entre tiempo(RunTime) y año(YEAR)
correlacion = df_limpio['YEAR_limpio'].corr(df_limpio['RunTime'])
print(correlacion)
```

```python
#Crear columna YEAR.
df_limpio['YEAR_limpio'] = df_limpio['YEAR'].astype(str).str.replace("(", "").str.replace(")", "").str.replace("I", "").astype(float)
```

```python
#Crear columna gross_float para poder usar para el calculo de otros indicadores
df_limpio['Gross_Float'] = df_limpio['Gross'].str.replace("M", "").str.replace("$","").astype(float).mul(1000000)
```

```python
df_limpio.head()
```

```python
# Preprocesamiento de columna GENRE limpiando saltos de línea, corchetes, comillas y barras invertidas
df_limpio['GENRE'] = df_limpio['GENRE'].astype(str).str.replace("\n", "").str.replace("[", "").str.replace("]", "").str.replace("'", "").str.replace('"', '').str.replace("\\", "").str.strip().str.split(", ")

# Explotamos la columna para que cada género tenga su propia fila
df_generos = df_limpio.explode('GENRE')

# Contamos la frecuencia de cada género
conteo = df_generos['GENRE'].value_counts()
print(conteo)
```
