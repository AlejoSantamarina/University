```python
#Mismo ejercicio pero con pequeñas modificaciones, por lo tanto resuelvo.
import pandas as pd
from ydata_profiling import ProfileReport
archivo = "/home/alejo/Projects/University/Fundamentos-de-la-Ciencia-de-Datos/TP2/catalogo_streaming_argentina_2024.xlsx"
df = pd.read_excel(archivo)
reporte = ProfileReport(df, title="Reporte", explorative=True)
reporte.to_file("reporte_catalogo_streaming_argentina.html")
```

```python
#Correlación entre variables.
import numpy as np

variable_1 = "imdbScore"
variable_2 = "tmdbScore"

datos_columna1 = df[variable_1]
datos_columna2 = df[variable_2]

coeficiente_correlacion = datos_columna1.corr(datos_columna2)
print(coeficiente_correlacion)
```

```python
#Generar scatter plots.
import matplotlib.pyplot as plt

plt.figure(figsize=(7,5))
plt.scatter(df[variable_1], df[variable_2], alpha=0.6, color="blue")

plt.xlabel(variable_1)
plt.ylabel(variable_2)
plt.title(f"Scatter plot: relación entre {variable_1} y {variable_2}")

plt.grid(True, linestyle="--", alpha=0.5)

plt.show()
```

```python
#DataFrame ordenado por plataformas con mas titulos
plataformas_con_mas_peliculas = df["platform"].value_counts().reset_index()
print(plataformas_con_mas_peliculas)
```
