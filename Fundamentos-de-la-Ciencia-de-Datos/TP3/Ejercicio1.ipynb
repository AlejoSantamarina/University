```python
#Generar reporte explorativo de obesidad.
import pandas as pd
from ydata_profiling import ProfileReport
df = pd.read_csv("/home/alejo/Projects/University/Fundamentos-de-la-Ciencia-de-Datos/Archivos/obesity_1.csv")
reporte = ProfileReport(df, title = "Reporte", explorative = True)
reporte.to_file("reporte_obesidad.html")
```

```python
#Calculo de correlación entre variables e impresión de scatter plots.
import numpy as np

variable_1 = "Height"
variable_2 = "Weight"

datos_columna1 = df[variable_1]
datos_columna2 = df[variable_2]

coeficiente_correlacion = np.corrcoef(datos_columna1, datos_columna2)[0, 1]
print(coeficiente_correlacion)
```

```python
#Generar scatter plots.
import matplotlib.pyplot as plt

plt.figure(figsize=(7,5))
plt.scatter(df[variable_1], df[variable_2], alpha=0.6, color="blue")

plt.xlabel(variable_1)
plt.xlabel(variable_2)
plt.title(f"Scatter plot: relación entre {variable_1} y {variable_2}")

plt.grid(True, linestyle="--", alpha=0.5)

plt.show()
```
