```python
class Rectangulo:

    def __init__(self, ancho, alto):
        self.ancho = ancho
        self.alto = alto

    def area(self):
        return int(self.ancho*self.alto)

    def perimetro(self):
        return int(self.ancho*2 + self.alto*2)

rectangulo1 = Rectangulo(8,5)
print(f"El area del rectangulo 1 es: {rectangulo1.area()}")
print(f"El perimetro del rectangulo 1 es: {rectangulo1.perimetro()}")
```

```python
class Cuadrado(Rectangulo):

    def __init__(self, lado):
        super().__init__(lado, lado)

cuadrado1 = Cuadrado(4)

print(f"El area del cuadrado es: {cuadrado1.area()}")
print(f"El perimetro del cuadrado es: {cuadrado1.perimetro()}")
```
