```python
def esPalindroma(palabra):
    i = 0
    j = len(palabra) - 1
    while palabra[i] == palabra[j] and i < j:
        i += 1
        j -= 1
    if(i >= j):
        return ("Es palindroma.")
    else: 
        return ("No es palindroma.")

print(esPalindroma(input("Ingrese la palabra:")))
```
